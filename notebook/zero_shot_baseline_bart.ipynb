```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification",model="facebook/bart-large-mnli")
```

```python
import json

with open("../data/processed/test_data.json") as f:
    test_samples = json.load(f)

labels = ["sarcastic", "not sarcastic"]

results = []
total = len(test_samples)
for i, sample in enumerate(test_samples):  
    print(f"Processed {i + 1} / {total}")
    prediction = classifier(sample["text"], labels)
    results.append({
        "text": sample["text"],
        "label": prediction["labels"][0],
        "scores": prediction["scores"]
    })
```

```python
with open("../data/processed/zero_shot_predictions.json", "w") as f:
    json.dump(results, f, indent=2)

print("Zero-shot predictions saved.")
```

```python
from sklearn.metrics import accuracy_score, f1_score, classification_report

with open("../data/processed/test_data.json") as f:
    test_samples = json.load(f)

with open("../data/processed/zero_shot_predictions.json") as f:
    predictions = json.load(f)

true_labels = [sample["label"] for sample in test_samples]
pred_labels = [1 if pred["label"] == "sarcastic" else 0 for pred in predictions]

acc = accuracy_score(true_labels, pred_labels)
f1 = f1_score(true_labels, pred_labels, average="weighted")  
report = classification_report(true_labels, pred_labels, target_names=["not sarcastic", "sarcastic"])

print(f"Accuracy: {acc:.4f}")
print(f"Weighted F1 Score: {f1:.4f}")
print("\nDetailed Classification Report:\n", report)
```
